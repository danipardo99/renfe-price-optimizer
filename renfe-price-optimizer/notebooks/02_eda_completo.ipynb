**Creacion de variables temporales para fechas**

```python
df["departure_month"] = df["departure"].dt.month
df["departure_weekday"] = df["departure"].dt.day_name()
df["departure_hour"] = df["departure"].dt.hour
df["is_weekend"] = (
    df["departure"].dt.weekday >= 5
).astype(int)

df["insert_hour"] = df["insert_date"].dt.hour

df["days_in_advance"] = (
    df["departure"] -
    df["insert_date"]
).dt.days


def temporada(mes):
    if mes in [12, 1, 2]:
        return "Invierno"
    elif mes in [3, 4, 5]:
        return "Primavera"
    elif mes in [6, 7, 8]:
        return "Verano"
    else:
        return "Otoño"

df["season"] = df["departure_month"].apply(temporada)
```

```python
#Crear el diccionario de normalización para ciudades con tildes y sin tildes
city_mapping = {
    "CORDOBA": "CÓRDOBA",
    "LEON": "LEÓN",
    "MALAGA": "MÁLAGA",
    "CASTELLO": "CASTELLON"
}


# Normalizar ciudades de origen
df["origin"] = df["origin"].replace(city_mapping)

# Normalizar ciudades de destino
df["destination"] = df["destination"].replace(city_mapping)
```

```python
# Diccionario de normalización de los tipos de tren

vehicle_type_mapping = {
    "Intercity": "INTERCITY",
    "REG.EXP.": "R. EXPRES"
}

#Reemplazar valores de tipo vehiculo
df["vehicle_type"] = df["vehicle_type"].replace(vehicle_type_mapping)

print(sorted(df["vehicle_type"].unique()))
```

```python
# Diccionario de normalización de los tipos de tren
#Se dejaran grupos por AVE, AVANT, MD y LD, ya que son los tipos de trenes más representativos de Renfe.
vehicle_type_mapping = {

    # Grupo AVE
    "AVE-AVE": "AVE",
    "AVE-MD": "AVE",
    "AVE-LD": "AVE",
    "AVE-AVANT": "AVE",
    "AVE-TGV": "AVE",

    # Grupo AVANT
    "AVANT-AVE": "AVANT",
    "AVANT-LD": "AVANT",
    "AVANT-MD": "AVANT",

    # Grupo MD
    "MD-AVE": "MD",
    "MD-LD": "MD",
    "MD-AVANT": "MD",

    # Grupo LD
    "LD-AVE": "LD",
    "LD-MD": "LD",
    "LD-AVANT": "LD"
}

df["vehicle_type"] = df["vehicle_type"].replace(vehicle_type_mapping)

print(sorted(df["vehicle_type"].unique()))
```

```python
# Diccionario de normalización de las clases de tren
#Se dejaran grupos por Turista, Turista Plus y Preferente, ya que son las clases de trenes más representativos de Renfe.

vehicle_class_mapping = {
    # Corrección de formato y agrupación de Turista
    "TuristaSólo plaza H": "Turista",
    "Turista Sólo plaza H": "Turista",
    "Turista con enlace": "Turista",

    # Agrupación de Turista Plus
    "Turista PlusSólo plaza H": "Turista Plus",
    "Turista Plus Sólo plaza H": "Turista Plus",
    "Turista - Turista Plus": "Turista Plus",
    "Turista Plus - Turista": "Turista Plus",

    # Agrupación de Preferente
    "PreferenteSólo plaza H": "Preferente",
    "Preferente Sólo plaza H": "Preferente"
}


df["vehicle_class"] = df["vehicle_class"].replace(vehicle_class_mapping)

print(sorted(df["vehicle_class"].dropna().unique()))

```

```python
# # Identificar posibles outliers mediante el método IQR

Q1 = df["price"].quantile(0.25)

Q3 = df["price"].quantile(0.75)

IQR = Q3 - Q1

limite_inferior = Q1 - 1.5 * IQR

limite_superior = Q3 + 1.5 * IQR

print("Límite inferior:", limite_inferior)

print("Límite superior:", limite_superior)
```

```python
#porcentaje de posibles outlayers en el precio de los billetes
outliers = df[
    (df["price"] < limite_inferior) |
    (df["price"] > limite_superior)
]

print("Número de posibles outliers:", len(outliers))

print("Porcentaje:",
      round(len(outliers) / len(df) * 100, 3),
      "%")
```

```python

#coeficiente de correlacion, numeros entre 1 y -1, si es cercano a 1 hay correlacion positiva, si es cercano a -1 hay correlacion negativa, si es cercano a 0 no hay correlacion
correlacion = df["precio"].corr(df["duracion"])

print(f"Correlación Precio-Duración: {correlacion:.4f}")

# existe una relación negativa; es una relación débil;

#la duración no pesa mucho en el precio del billete, lo que indica que el precio no está directamente relacionado con la duración del viaje.
# Esto puede deberse a que otros factores, como la clase del billete, la tarifa y el tipo de tren, tienen un impacto más significativo en el precio final.
```

```python
#Tomamos unamuestra aleatoria de 100.000 registros para visualizar la relación entre precio y duración del viaje.

muestra = df.sample(100000, random_state=42)

import matplotlib.pyplot as plt

plt.figure(figsize=(9,6))

plt.scatter(
    muestra["duracion"],
    muestra["precio"],
    alpha=0.15,
    s=4
)

plt.title("Precio vs duración del viaje")

plt.xlabel("Duración (horas)")

plt.ylabel("Precio (€)")

plt.grid(alpha=0.3)

plt.show()

```

```python
#Dibujamos la línea de tendencia para ver si existe una relación lineal entre precio y duración del viaje. Para ello,
# utilizamos la función polyfit de NumPy para calcular los coeficientes de la recta de regresión lineal.




muestra = df.sample(100000, random_state=42)

x = muestra["duracion"]

y = muestra["precio"]

coef = np.polyfit(x, y, 1)

recta = np.poly1d(coef)

plt.figure(figsize=(9,6))

plt.scatter(
    x,
    y,
    alpha=0.12,
    s=4
)

plt.plot(
    np.sort(x),
    recta(np.sort(x)),
    color="red",
    linewidth=3,
    label="Tendencia"
)

plt.title("Relación entre duración y precio")

plt.xlabel("Duración (horas)")

plt.ylabel("Precio (€)")

plt.legend()

plt.grid(alpha=0.3)

plt.show()

#La recta descendente confirma la correlación negativa.
```

**Analisis de correlación entre variables numércias**

```python
variables_numericas = [
    "precio",
    "duracion",
    "dias_anticipacion",
    "hora_salida",
    "hora_captura"
]

df[variables_numericas].head()

```

```python
#Matriz de correlación entre las variables numéricas
corr = df[variables_numericas].corr()

corr.round(3)
```

```python
#Mapa de calor de la matriz de correlación

import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.imshow(corr,
           cmap="coolwarm",
           vmin=-1,
           vmax=1)

plt.colorbar(label="Correlación")

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=45,
    ha="right"
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

for i in range(len(corr)):
    for j in range(len(corr)):
        plt.text(
            j,
            i,
            f"{corr.iloc[i,j]:.2f}",
            ha="center",
            va="center",
            fontsize=10,
            color="black"
        )

plt.title("Matriz de correlación de variables numéricas")

plt.tight_layout()

plt.show()
```
